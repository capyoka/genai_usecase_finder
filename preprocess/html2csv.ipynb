```python
from bs4 import BeautifulSoup
import csv

INPUT_FILE = "raw/notion_table.html"
OUTPUT_FILE = "pr/notion_table.csv"

# ==== 1. HTMLファイルを読み込む ====
# 例: 同じフォルダに notion_table.html を保存しておく
with open(INPUT_FILE, "r", encoding="utf-8") as f:
    html = f.read()

# ==== 2. BeautifulSoup で解析 ====
soup = BeautifulSoup(html, "html.parser")

# ==== 3. 各行（row）を抽出 ====
rows = soup.select(".notion-table-view-row")

# ==== 4. CSV出力用リストを作成 ====
data = []

for row in rows:
    # 各セルのテキストを抽出
    cells = row.select(".notion-table-view-cell")
    row_data = []

    for cell in cells:
        # <a>タグがある場合はリンクも含めて取得
        link = cell.find("a", href=True)
        if link:
            text = link.get_text(strip=True)
            href = link["href"]
            row_data.append(f"{text} ({href})")
        else:
            # 通常のテキストを取得
            text = cell.get_text(separator=" ", strip=True)
            row_data.append(text)

    data.append(row_data)

# ==== 5. CSVとして保存 ====


with open(OUTPUT_FILE, "w", newline="", encoding="utf-8-sig") as f:
    writer = csv.writer(f)
    # ヘッダー（Notionの列構造に合わせて任意で設定）
    header = [
        "カテゴリ",
        "組織名",
        "タイトル",
        "概要",
        "テーマ",
        "リンク",
        "担当者"
    ]
    writer.writerow(header)

    # 各行データを書き込み
    for row in data:
        writer.writerow(row)

print(f"✅ {OUTPUT_FILE} に {len(data)} 件のレコードを書き出しました。")
```
